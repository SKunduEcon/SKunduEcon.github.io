# US payroll employment recovery after recessions
A small reproducible chart using the recovery-month figures reported by RAND (July 2022). The values are descriptive episode summaries, not estimates of a causal effect.

**Source:** RAND, [Recovered but not whole: US jobs rebounded after the pandemic, but employment recovery differs across recessions](https://www.rand.org/pubs/commentary/2022/07/recovered-but-not-whole-us-jobs-rebounded.html).

```python
import matplotlib.pyplot as plt

episodes = ['1990–91', '2001', '2007–09', '2020']
months = [31, 46, 77, 29]

fig, ax = plt.subplots(figsize=(8, 4.5))
bars = ax.bar(episodes, months)
ax.set_title('Months for US payroll employment to regain its pre-recession level')
ax.set_ylabel('Months')
ax.set_ylim(0, 90)
ax.bar_label(bars, labels=[f'{v} months' for v in months], padding=4)
ax.spines[['top', 'right']].set_visible(False)
fig.tight_layout()
plt.show()
```
